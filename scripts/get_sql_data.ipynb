```python
#Libraries
import pandas as pd
import mysql.connector

#Conection to db, ssh tunneling is required to connect to the database
conection = mysql.connector.connect(
    host="127.0.0.1", port=3306,
    user="read", password="pass", #REMEMBER TO REMOVE LATER
    database="cowrie"
)
```

```python
#UNIQUE IPs THAT TRIED TO CONNECT TO THE HONEYPOT

query_unique_ips ="""
        SELECT ip, COUNT(*) AS intentos
        FROM sessions
        GROUP BY ip
        ORDER BY intentos DESC
    """
unique_ips = pd.read_sql(query_unique_ips, con=conection)

unique_ips.to_csv('unique_ips.csv', index=False) #Backup
```

```python
#COMMANDS THAT WERE EXECUTED IN EACH SESSION

commands_per_session = pd.read_sql("SELECT * FROM input", conection)
commands_per_session.to_csv("inputs.csv", index=False) #BACKUP
```

```python
#USER AND PASSWORDS THAT WERE USED TO TRY TO CONNECT TO THE HONEYPOT

users_passwords = pd.read_sql("SELECT * FROM auth", conection)
users_passwords.to_csv("users_passwords.csv", index=False) #BACKUP
```
